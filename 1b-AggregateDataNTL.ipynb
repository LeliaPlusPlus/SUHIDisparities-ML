# Create Study Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
uhi = pd.read_csv('data/UHI.csv')
cej = pd.read_csv('data/CEJST-v1.csv', low_memory=False)
```

```python
cej = cej[['Census tract 2010 ID', #'County Name', 'State/Territory',  
           'Is low income?', 'Percent of individuals < 100% Federal Poverty Line']]

cols = {#'County Name': 'County', 'State/Territory': 'State', 
        #'Percent Black or African American alone': 'Black / African American', 'Percent Asian': 'Asian',  
        #'Percent White': 'White', 'Percent Hispanic or Latino': 'Hispanic / Latino', 
        'Percent of individuals < 100% Federal Poverty Line': 'Below Poverty'}

cej.rename(columns=cols, inplace=True)
```

```python
uhi = uhi[['GEOID', 'County', 'State', 'Area', 'Pop2020', 'BSA2020', 'NDBI2020', 'NDVI2020', 'day2020', 'DevelopedArea', 
           'Black2020', 'Hispanic2020', 'White', 'Income']]

cols = {'BSA2020': 'BSA', 'Pop2020': 'Population', 'NDBI2020': 'NDBI', 
        'NDVI2020': 'NDVI', 'day2020': 'SUHI', 'DevelopedArea': 'Developed Area', 
        'Black2020': 'Black', 'Hispanic2020': 'Hispanic', 'Income': 'Median Income'}

uhi.rename(columns=cols, inplace=True)
```

```python
len(uhi)
```

```python
uhi.head()
```

```python
data = uhi.merge(cej, how='left', left_on='GEOID', right_on='Census tract 2010 ID')
```

```python
data.head()
```

```python
data['Black']    = data['Black'] * 100
data['Hispanic'] = data['Hispanic'] * 100
data['White']    = data['White'] * 100
```

## Add urban information

```python
urban_df = pd.read_csv('data/Census_UHI_US_Urbanized_climzone_metadata.csv')
uhi_df   = pd.read_csv('data/Census_UHI_US_Urbanized_vegetation_albedo_CUHI_SUHI.csv')
```

```python
uhi_df.head()
```

```python
uhi_df['DEM'] = uhi_df['DEM_urb_CT_act'] - uhi_df['DEM_rur']
```

```python
urban_df.head()
```

```python
data = data.merge(uhi_df[['Census_geoid', 'Urban_name', 'DEM']], 
                  how='inner', left_on='GEOID', right_on='Census_geoid')
```

```python
data = data.merge(urban_df[['Urban_name', 'Clim_zone', 'Coastal?']], 
                  how='right', left_on='Urban_name', right_on='Urban_name')
```

```python
data.rename(columns={'Clim_zone': 'Climate Zone'}, inplace=True)
```

## Normalize demographic data by city as percentiles

```python
data['Median Income Percentile'] = data.groupby('Urban_name')['Median Income'].rank(pct=True)
data['Black Percentile'] = data.groupby('Urban_name')['Black'].rank(pct=True)
data['Hispanic Percentile'] = data.groupby('Urban_name')['Hispanic'].rank(pct=True)
data['White Percentile'] = data.groupby('Urban_name')['White'].rank(pct=True)
```

```python
data.head()
```

```python
len(data)
```

```python
data['NDVI'] = data['NDVI'] * 0.0001
```

```python
data['BSA'] = data['BSA'] * 0.001
```

```python
d_stats = data[['BSA', 'NDBI',
                'NDVI', 'DEM',
                'Black', 'Hispanic', 'White', 
                'Below Poverty', 'SUHI']].describe().round(4)
d_stats.T
```

```python
missing = pd.DataFrame({'Total Missing': data.isna().sum(), 'Percent Missing': (data.isna().sum() / len(data)) * 100 })
missing
```

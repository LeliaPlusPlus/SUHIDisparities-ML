## Data

```python
df_ntl = pd.read_csv('data/data_ntl.csv')
df_sue = pd.read_csv('data/data_sue.csv')
```

```python
old_cols = ['BSA', 'NDBI', 'NDVI', 'DEM']
new_cols = ['\u0394BSA', '\u0394NDBI', '\u0394NDVI', '\u0394DEM']

df_ntl.rename(columns=dict(zip(old_cols, new_cols)), inplace=True)
df_sue.rename(columns=dict(zip(old_cols, new_cols)), inplace=True)
```
